```python
# import libraries we need for matrices, table, and graphs
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# creates labels for the movie colums and user rows
movies = ["Action 1", "Action 2", "Action 3",
          "Romance 1", "Romance 2", "Romance 3"]

users = ["Ana", "Ben", "Cal", "Dee", "Eve", "Fay"]
```

```python
# create the toy rating matrix
# np.nan means that the user didn't rate that movie
toy = np.array([
    [5, 4, 5, 1, np.nan, 2],       # Ana
    [4, np.nan, 5, 2, 1, 1],       # Ben
    [5, 5, np.nan, 1, 2, np.nan],  # Cal
    [1, 2, 1, 5, 4, np.nan],       # Dee
    [np.nan, 1, 2, 4, 5, 5],       # Eve
    [2, 1, np.nan, 5, np.nan, 4]   # Fay
])
```

```python
# Turn the matrix into a labeled table so it is easier to read
toy_table = pd.DataFrame(toy, index=users, columns=movies)

toy_table
```

```python
# Count how many ratings are missing in the toy matrix
missing = np.isnan(toy).sum()

print("Missing entries:", missing, "out of", toy.size)
```

Ana, Ben, and Cal usually like action movies more than romance movies. Dee, Eve, and Fay usually prefer romance movies. I can use ratings from similar users to estimate the missing values.

```python
# Import tools for downloading and unzipping the MovieLens dataset
import os
import urllib.request
import zipfile
import ssl

# Create an SSL context for the download
ssl_context = ssl._create_unverified_context()

# Download the dataset only if it is not already in Colab
if not os.path.exists("ml-100k"):
    url = "https://files.grouplens.org/datasets/movielens/ml-100k.zip"

    # Download the zip file
    with urllib.request.urlopen(url, context=ssl_context) as response:
        with open("ml-100k.zip", "wb") as file:
            file.write(response.read())

    # Unzip the downloaded dataset
    zipfile.ZipFile("ml-100k.zip").extractall()

print("MovieLens dataset is ready!")
```

```python
# Load the ratings data into a table
ratings = pd.read_csv(
    "ml-100k/u.data",
    sep="\t",
    names=["user", "item", "rating", "time"]
)

ratings.head()
```

```python
# Count the total ratings, unique users, and unique movies
print("Number of ratings:", len(ratings))
print("Number of users:", ratings["user"].nunique())
print("Number of movies:", ratings["item"].nunique())
```

```python
# Count how many times each star rating appears
star_counts = ratings["rating"].value_counts().sort_index()

star_counts
```

```python
# Calculate the average rating in the dataset
average_rating = ratings["rating"].mean()

print("Average rating:", round(average_rating, 3))
```

```python
# Create a bar chart showing how often each star rating appears
star_counts.plot(kind="bar", color="steelblue")

plt.xlabel("Stars")
plt.ylabel("Number of ratings")
plt.title("MovieLens 100K: How people rate")
plt.show()
```

```python
# Convert the raw ratings table into a user-by-movie matrix
R = ratings.pivot_table(
    index="user",
    columns="item",
    values="rating")
```

```python
# Show the number of rows and columns in the ratings matrix
print("Matrix shape:", R.shape)

# Calculate the fraction of entries that are missing
fraction_missing = R.isna().to_numpy().mean()

print("Fraction missing:", round(fraction_missing, 4))
```

```python
# Count how many movies each user rated
per_user = ratings.groupby("user").size()

print("Least active user:", per_user.min())
print("Median user:", int(per_user.median()))
print("Most active user:", per_user.max())
```

```python
# Load movie ID numbers and movie titles
titles = pd.read_csv(
    "ml-100k/u.item",
    sep="|",
    encoding="latin-1",
    header=None,
    usecols=[0, 1],
    names=["item", "title"]
)

titles.head()
```

```python
# Find the number of ratings and average rating for each movie
stats = ratings.groupby("item")["rating"].agg(["count", "mean"])

# Add the movie titles to the table
stats = stats.join(titles.set_index("item"))

# Show the five movies with the most ratings
top_five = stats.sort_values("count", ascending=False).head(5)

top_five.round(2)
```

Movie ratings are influenced by a smaller number of hidden preferences, such as action vs. romance, comedy vs. drama, and older vs. newer movies. Because only a few factors may explain many ratings, the full ratings matrix can be approximated by a low-rank matrix.

```python
# Add a seventh user who likes action movies and dislikes romance movies
new_user = [5, 4, np.nan, 1, 2, np.nan]

# Add the new user to the toy matrix
toy_7 = np.vstack([toy, new_user])

# Add the new user's name
users_7 = users + ["Gina"]

# Display the updated matrix
toy_7_table = pd.DataFrame(toy_7, index=users_7, columns=movies)

toy_7_table
```

```python
# Find the fraction of missing entries in the toy matrix
fraction_missing_toy = np.isnan(toy).mean()

print("Fraction missing:", fraction_missing_toy)
```

```python
# Count how many times each star rating appears
star_counts = ratings["rating"].value_counts().sort_index()

# Find the least common rating
least_common_rating = star_counts.idxmin()
least_common_count = star_counts.min()

print("Least common star rating:", least_common_rating)
print("Number of occurrences:", least_common_count)
```

The least common star rating is 1 star, and it appears 6,110 times in the dataset.

```python
# Find the average rating given by each user
user_averages = ratings.groupby("user")["rating"].mean()

# Find the user with the highest average rating
highest_user = user_averages.idxmax()
highest_average = user_averages.max()

print("User with highest average rating:", highest_user)
print("Highest average rating:", round(highest_average, 2))
```

User 849 has the highest average rating of 4.87 stars. We should be cautious about calling this user the “most positive” because users rated different numbers of movies, so a high average may come from a smaller or less representative set of ratings.

The ratings table stores one individual rating per row, including the user ID, movie ID, rating, and time. The matrix R reorganizes the same information so that each row represents one user and each column represents one movie. In R, each entry shows the rating that user gave that movie, and missing ratings appear as NaN.

If the rating matrix had no missing values, SVD could be computed because every entry would be a number. However, SVD would still not automatically solve the recommendation problem because we would still need to decide how to use the lower-rank approximation to make useful predictions and recommendations for users.

1. Action preference: A high value for a user means they strongly like action movies. A high value for a movie means the movie has a lot of action.  
2. Comedy preference: A high value for a user means they enjoy funny movies. A high value for a movie means the movie has a strong comedy element.  
3. Preference for newer movies: A high value for a user means they prefer newer films. A high value for a movie means the movie is relatively recent.
